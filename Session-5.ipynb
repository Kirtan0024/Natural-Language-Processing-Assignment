## Task 3: Sentiment Analysis Using Hugging Face

```python
from transformers import pipeline

sentiment = pipeline("sentiment-analysis")

reviews = [
    "The food was delicious and delivery was very fast.",
    "The food was cold and the delivery was very late.",
    "Good taste and reasonable price, I really enjoyed the meal."
]

for review in reviews:
    result = sentiment(review)
    print("Review:", review)
    print("Sentiment:", result[0]["label"])
    print("Score:", result[0]["score"])
    print()
```

## Task 5: Basic Attention Mechanism

```python
import numpy as np 

def attention_example(sentences):
    words = []

    for sentence in sentences:
        words.extend(sentence.lower().split())

    words = list(dict.fromkeys(words))

    scores = []

    for word in words:
        score = 0

        for sentence in sentences:
            if word in sentence.lower().split():
                score += 1

        scores.append(score)

    for word, score in zip(words, scores):
        print(word, ":", score)

sentences = [
    "I love pizza",
    "Pizza is very tasty",
    "I order pizza online"
]

attention_example(sentences)
```
