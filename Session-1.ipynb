## Task 4: Sentiment Checker

```python
review = input("Enter your movie review: ")
review = review.lower()

if "good" in review:
    print("Positive")
elif "bad" in review:
    print("Negative")
else:
    print("Neutral")
```

## Task 5: Spam Filter

```python
def is_spam(email_text):
    email_text = email_text.lower()

    if "win" in email_text or "free" in email_text:
        return True
    else:
        return False

print(is_spam("You can win a free prize"))
print(is_spam("Your order has been delivered"))
```
