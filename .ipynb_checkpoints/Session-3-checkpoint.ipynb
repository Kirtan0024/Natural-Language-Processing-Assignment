## Task 1: Create Vocabulary from Instagram Captions

```python
import string

caption1 = "Loving this beautiful sunset today!"
caption2 = "Amazing food and great vibes!"
caption3 = "Beautiful day with amazing friends!"

captions = [caption1, caption2, caption3]

vocabulary = set()

for caption in captions:
    caption = caption.lower()
    caption = caption.translate(str.maketrans("", "", string.punctuation))
    words = caption.split()
    vocabulary.update(words)

vocabulary = sorted(vocabulary)

print(vocabulary)
```

## Task 2: Bag of Words using CountVectorizer

```python
from sklearn.feature_extraction.text import CountVectorizer

reviews = [
    "Loved the pizza",
    "Service was slow",
    "Great ambience"
]

vectorizer = CountVectorizer()
matrix = vectorizer.fit_transform(reviews)

print(vectorizer.get_feature_names_out())
print(matrix.toarray())
```

## Task 3: Word Frequency in WhatsApp Messages

```python
messages = [
    "I am going to the market",
    "The market is very busy",
    "I am going home"
]

stopwords = ["the", "is", "at", "am", "to"]

word_count = {}

for message in messages:
    words = message.lower().split()

    for word in words:
        if word not in stopwords:
            if word in word_count:
                word_count[word] += 1
            else:
                word_count[word] = 1

for word, count in word_count.items():
    print(word, ":", count)
```

## Task 4: TF-IDF using TfidfVectorizer

```python
from sklearn.feature_extraction.text import TfidfVectorizer

reviews = [
    "The quality of this product is excellent",
    "The product quality is good",
    "Poor quality and bad packaging"
]

vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(reviews)

words = vectorizer.get_feature_names_out()
quality_index = list(words).index("quality")

for i in range(len(reviews)):
    print("Review", i + 1, ":", tfidf_matrix[i, quality_index])
```

## Task 5: Bag of Words vs TF-IDF

```python
# Bag of Words only counts how many times a word appears in a document, while TF-IDF gives more importance to words that are useful in a 
# particular document.
# Bag of Words treats common words and important words more similarly, while TF-IDF reduces the importance of words that appear in many documents.

# A real-world example is a search feature in an online shopping app like Flipkart. If a user searches for "best camera quality",
# TF-IDF can help identify reviews or product descriptions that are more relevant to the search because it gives more importance
# to specific words such as "camera" and "quality" when they are less common across all documents.
```
