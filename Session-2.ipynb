## Task 1: Convert Instagram Bio to Lowercase

```python
def lowercase_bio(bio):
    return bio.lower()

bio = input("Enter your Instagram bio: ")
print(lowercase_bio(bio))
```

## Task 2: Remove Punctuation from Flipkart Review

```python
import string

def remove_punctuation(review):
    cleaned_text = ""

    for char in review:
        if char not in string.punctuation:
            cleaned_text += char

    return cleaned_text

review = input("Enter your product review: ")
print(remove_punctuation(review))
```

## Task 4: Remove Stopwords from WhatsApp Message

```python
import nltk
from nltk.corpus import stopwords

nltk.download("stopwords")

def remove_stopwords(message):
    stop_words = set(stopwords.words("english"))
    words = message.split()
    cleaned_words = []

    for word in words:
        if word.lower() not in stop_words:
            cleaned_words.append(word)

    return " ".join(cleaned_words)

message = input("Enter your WhatsApp message: ")
print(remove_stopwords(message))
```

## Task 5: Porter Stemmer for Trending Hashtags

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

hashtags = ["#playing", "#dancing", "#eating"]

for hashtag in hashtags:
    word = hashtag.replace("#", "")
    stemmed_word = stemmer.stem(word)
    print(hashtag, "->", "#" + stemmed_word)
```
